**Q02. Given the following data, create a line chart with the following properties:**

```python
week_day = ['SUN','MON','TUE','WED','THU','FRI','SAT']
high_temp = [88, 97, 95, 98, 100, 101, 93]
low_temp = [77, 80, 79, 73, 80, 81, 78]
avg_temp = [82, 88, 87, 85, 90, 91, 86]
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

plt.plot(week_day, high_temp, color='red', linewidth=2, linestyle='--', marker='o', alpha=0.8, label='High Temp')
plt.plot(week_day, low_temp, color='blue', linewidth=2, linestyle=':', marker='s', alpha=0.8, label='Low Temp')
plt.plot(week_day, avg_temp, color='green', linewidth=2, linestyle='-', marker='^', alpha=0.8, label='Avg Temp')

plt.title('Weekly Temperature')
plt.xlabel('Week Day')
plt.ylabel('Temperature in Fahrenheit')

plt.legend(loc='lower right')
plt.grid(True)

plt.show()
```

**Q03. Use the data below, create a stacked bar chart. Use the output image as reference**

```python
import numpy as np
import matplotlib.pyplot as plt

years = range(2015,2020)
class_counts = {
    'Face to Face':np.array([47, 50, 48, 40, 35]),
    'Online': np.array([29, 28, 40, 45, 50]),
}

plt.figure(figsize=(10, 6))

bar_face_to_face = plt.bar(years, class_counts['Face to Face'], label='Face to Face', color='blue')
bar_online = plt.bar(years, class_counts['Online'], bottom=class_counts['Face to Face'], label='Online', color='red')

plt.xlabel('Year')
plt.ylabel('Number of Classes')
plt.title('Class Modality Over Years')
plt.xticks(years)
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Add labels to the bars
for bar1, bar2 in zip(bar_face_to_face, bar_online):
    plt.text(bar1.get_x() + bar1.get_width() / 2, bar1.get_height() / 2, str(int(bar1.get_height())), ha='center', va='center', color='black', fontsize=9)
    plt.text(bar2.get_x() + bar2.get_width() / 2, bar1.get_height() + bar2.get_height() / 2, str(int(bar2.get_height())), ha='center', va='center', color='black', fontsize=9)

plt.show()
```

**Q04. Load dataset 'iris' from Seaborn and create a stacked Histogram chart to see the distribution of sepal length. Use the output image as reference.**

```python
import seaborn as sns
flowers_df = sns.load_dataset("iris")
print(flowers_df)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

flowers_df = sns.load_dataset('iris')

# Get sepal length for each species
setosa_sepal_length = flowers_df[flowers_df['species'] == 'setosa']['sepal_length']
versicolor_sepal_length = flowers_df[flowers_df['species'] == 'versicolor']['sepal_length']
virginica_sepal_length = flowers_df[flowers_df['species'] == 'virginica']['sepal_length']

plt.figure(figsize=(10, 6))
plt.hist(
    [setosa_sepal_length, versicolor_sepal_length, virginica_sepal_length],
    bins=10, # Number of bins for the histogram
    stacked=True, # This makes it a stacked histogram
    label=['setosa', 'versicolor', 'virginica'],
    color=['blue', 'red', 'green'] # Custom colors
)

plt.title('Stacked Histogram of Sepal Length by Species')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Frequency')
plt.legend(title='Species')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

**Q08. Create a pie chart based on the data below to show class modality and its weight. Use the output image as reference.**

```python
modality = ['Face to Face', 'Online Anytime', 'Online with Schedule']
weight = [30, 50, 20]

explode = [0.1, 0, 0] # Explode the 'Face to Face' sector

plt.figure(figsize=(8, 8))
plt.pie(weight, labels=modality, autopct='%1.1f%%', startangle=90, colors=['blue', 'red', 'green'], explode=explode)
plt.title('Class Modality Distribution')
plt.axis('equal') # Equal aspect ratio ensures that pie is drawn as a circle.
plt.show()
```
